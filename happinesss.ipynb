# WORLD HAPPINESS REPORT 2021

## DESCRIPTION

The World Happiness Report is a leading publication on global well-being, aiming to measure and understand happiness across nations and provide insights for improving quality of life. It ranks over 150 countries based on how people evaluate their own lives on a 0–10 scale, using data primarily from the Gallup World Poll. The report also examines correlations between happiness and factors such as GDP per capita, healthy life expectancy, social support, freedom, generosity, and perceptions of corruption
