## DESCRIPTION

The World Happiness Report is a leading publication on global well-being, aiming to measure and understand happiness across nations and provide insights for improving quality of life. It ranks over 150 countries based on how people evaluate their own lives on a 0–10 scale, using data primarily from the Gallup World Poll. The report also examines correlations between happiness and factors such as GDP per capita, healthy life expectancy, social support, freedom, generosity, and perceptions of corruption

## BOX PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,3))
sns.boxplot(data=df,x='Score',color='skyblue')
plt.title("Distribution of score")
plt.xlabel("Score of countries")
plt.ylabel("Regional indicator")
plt.show()
```

#### INSIGHTS

* Shows the spread of “Score” values across countries using a boxplot.
* The median line indicates the central (typical) happiness score.
* The box (IQR) shows where most countries’ scores lie.
* Whiskers represent the overall range of scores (excluding outliers).
* Any points outside the whiskers indicate outliers (very high or low scores).
* Helps quickly understand variation and consistency in country scores.

```python
import matplotlib.pyplot as plt
df_num = df.select_dtypes(include=["number"])
for col in df_num.columns:
    plt.figure(figsize=(6,4))
    plt.boxplot(df_num[col].dropna())
    plt.title(col)
    plt.show()
```

#### INSIGHTS

* Creates boxplots for all numerical columns in the dataset.
* Helps compare the distribution of each numeric variable separately.
* Shows median, spread (IQR), and range for every column.
* Makes it easy to identify outliers in each feature.
* Useful for understanding data variability and consistency across variables

.Social Support

.Healthy Life Expectancy

.Freedom to Make Life Choices

Insight:
These are the strongest contributors to happiness among countries.

2. Moderate Correlation

.Generosity shows weak to moderate positive correlation with happiness.

.It influences happiness, but less strongly than GDP or health factors.

3. Negative Correlation

.Perceptions of Corruption generally shows a negative correlation with happiness.


Insight:
Countries with higher perceived corruption tend to have lower happiness scores.

4. High Inter-Correlation Between Predictors
